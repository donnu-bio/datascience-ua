## Фільтрація та предикати

Метод `.фільтрувати()` (`where`) залишає лише ті рядки, для яких умова істинна.

Загальна форма:

```python
таблиця.фільтрувати("Назва_колонки", умови.предикат(значення))
```

````python
листки.фільтрувати("Площа_мм2", умови.більше_ніж(30))
листки.фільтрувати("Вид", умови.дорівнює("Береза"))
листки.фільтрувати("Площа_мм2", умови.між(25, 50))
````

### Основні предикати `умови`

| Український псевдонім | Оригінал | Значення |
|---|---|---|
| `дорівнює(x)` | `equal_to(x)` | дорівнює `x` |
| `не_дорівнює(x)` | `not_equal_to(x)` | не дорівнює `x` |
| `більше_ніж(x)` | `above(x)` | більше за `x` |
| `менше_ніж(x)` | `below(x)` | менше за `x` |
| `не_менше_ніж(x)` | `above_or_equal_to(x)` | більше або дорівнює `x` |
| `не_більше_ніж(x)` | `below_or_equal_to(x)` | менше або дорівнює `x` |
| `між(a, b)` | `between(a, b)` | `a ≤ значення < b` |
| `між_включно(a, b)` | `between_or_equal_to(a, b)` | `a ≤ значення ≤ b` |
| `строго_між(a, b)` | `strictly_between(a, b)` | `a < значення < b` |
| `містить(s)` | `containing(s)` | рядок містить фрагмент `s` |
| `не_містить(s)` | `not_containing(s)` | рядок не містить `s` |
| `належить_до(A)` | `contained_in(A)` | значення є елементом `A` |

**Примітка:** `між(a, b)` не включає верхню межу `b`. Для включення обох меж використовуйте `між_включно(a, b)`.

### Поєднання умов

Для поєднання кількох критеріїв можна використовувати булеві масиви NumPy та оператори `&` (І), `|` (АБО), `~` (НЕ). Кожну умову беріть у дужки.

````python
# Листки з площею понад 25 мм² і шириною меншою за 60 мм
листки.фільтрувати(
    (листки.стовпець("Площа_мм2") > 25) &
    (листки.стовпець("Ширина_мм") < 60)
)
````

NumPy — бібліотека для обчислень із числовими масивами. У цьому довіднику українські псевдоніми згруповані в `np_ua`. Звичайний `np` також залишається доступним.

## Рядки

Рядки Python мають власні методи. Вони не є методами `datascience` чи NumPy.

````python
назва = "Carpinus betulus"

назва.split(" ")                 # розділити рядок на частини
" | ".join(["A", "B", "C"])      # об'єднати рядки через роздільник
назва.replace("betulus", "sp.")  # замінити фрагмент
назва.lower()                    # нижній регістр
назва.upper()                    # верхній регістр
назва.strip()                    # прибрати пробіли на краях
````

| Метод | Призначення |
|---|---|
| `рядок.split(роздільник)` | Розділити рядок на список |
| `роздільник.join(послідовність)` | Об'єднати рядки через роздільник |
| `рядок.replace(старий, новий)` | Замінити фрагмент |
| `рядок.lower()` | Перетворити на нижній регістр |
| `рядок.upper()` | Перетворити на верхній регістр |
| `рядок.strip()` | Видалити пробіли на початку й у кінці |
